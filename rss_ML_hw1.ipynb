```python
df_asia_select = df[df['origin'] == 'Asia'][['vehicle_weight', 'model_year']].head(7)
print(df_asia_select)
```

```python
X = df_asia_select.to_numpy()
X
```

```python
XTX = X.T.dot(X)
print("\nMatriz XTX:")
print(XTX)
```

```python
XTX_inv = np.linalg.inv(XTX)
print("Inversa de XTX:")
print(XTX_inv)
```

```python
y = [1100, 1300, 800, 900, 1000, 1100, 1200]
y = np.array(y)
print(y)
```

```python
w = XTX_inv.dot(X.T).dot(y)
print(w)
```

```python
suma_elem_w = np.sum(w)
print(suma_elem_w)
```
